```python
df_clean = df.copy()
```

```python
df_clean.isnull().sum()
df_clean['children'] = df_clean['children'].fillna(0)
```

```python
df_clean['country'] = df_clean['country'].fillna('Unknown')
df_clean['agent'] = df_clean['agent'].fillna(0)
```

```python
df_clean['company'] = df_clean['company'].fillna(0)
df_clean = df_clean.drop_duplicates()
```

```python
print(df_clean.duplicated().sum())
```

```python
df_clean['arrival_date'] = pd.to_datetime(
    df_clean['arrival_date_year'].astype(str) + '-' +
    df_clean['arrival_date_month'] + '-' +
    df_clean['arrival_date_day_of_month'].astype(str),
    format='%Y-%B-%d'
)
```

```python
df_clean.dtypes
df_clean.isnull().sum()
df_clean.shape
```

```python
"""We have three columns:
- Adults
- Children
- Babies
Let's combine them to calculate the total number of guests."""


df_clean['total_guests'] = (
    df_clean['adults'] +
    df_clean['children'] +
    df_clean['babies']
)
```

```python
"""Create Total Nights
We have two columns:
- stays_in_weekend_nights
- stays_in_week_nights
Let's combine them."""


df_clean['total_nights'] = (
    df_clean['stays_in_weekend_nights'] +
    df_clean['stays_in_week_nights']
)
```

```python
#Create Booking Month
df_clean['arrival_month'] = df_clean['arrival_date'].dt.month_name()
```

```python
#Create Arrival Year
df_clean['arrival_year'] = df_clean['arrival_date'].dt.year
```

```python
#Create Weekend Stay Indicator 
df_clean['weekend_stay'] = np.where(
    df_clean['stays_in_weekend_nights'] > 0,
    'Yes',
    'No'
)
```

```python
#Create Family Booking Indicator

df_clean['family_booking'] = np.where(
    (df_clean['children'] > 0) | (df_clean['babies'] > 0),
    'Yes',
    'No'
)
```

```python
#Create Booking Status 
df_clean['booking_status'] = np.where(
    df_clean['is_canceled'] == 1,
    'Canceled',
    'Not Canceled'
)
```

```python
#Check Newly Created Columns
df_clean[['total_guests',
          'total_nights',
          'arrival_month',
          'arrival_year',
          'weekend_stay',
          'family_booking',
          'booking_status']].head(10)
```

```python
#Hotel Booking Cancellation Analysis (Pandas)
#1. Cancellation by Hotel Type
df_clean.groupby('hotel')['is_canceled'].sum()
```

```python
#2. Cancellation Rate by Hotel Type

print((df_clean.groupby('hotel')['is_canceled'].mean() * 100).round(2))
```

```python
#3. Cancellation by Customer Type
df_clean.groupby('customer_type')['is_canceled'].sum().sort_values(ascending=False)
```

```python
#4. Cancellation Rate by Customer Type
(df_clean.groupby('customer_type')['is_canceled'].mean() * 100).round(2)
```

```python
#5. Cancellation by Deposit Type
df_clean.groupby('deposit_type')['is_canceled'].sum()
(df_clean.groupby('deposit_type')['is_canceled'].mean() * 100).round(2)
```

```python
#6. Cancellation by Market Segment
(df_clean.groupby('market_segment')['is_canceled'].mean() * 100).round(2).sort_values(ascending=False)
```

```python
#7. Cancellation Based on Lead Time
df_clean['lead_time_group'] = pd.cut(
    df_clean['lead_time'],
    bins=[-1, 30, 90, 180, 365, float('inf')],
    labels=['0-30 Days', '31-90 Days', '91-180 Days', '181-365 Days', '365+ Days']
)
(df_clean.groupby('lead_time_group', observed=True)['is_canceled'].mean() * 100).round(2)
```

```python
#8. Cancellation by Customer Segment and Hotel Type
(df_clean.groupby(['hotel', 'customer_type'])['is_canceled'].mean() * 100).round(2)
```

```python
#Revenue & Pricing Analysis (Pandas + NumPy)
#1. Average Daily Rate (ADR) by Hotel
df_clean.groupby('hotel')['adr'].mean().round(2)
```

```python
#2. Average ADR by Month
df_clean.groupby('arrival_month')['adr'].mean().round(2)
```

```python
#3. Average ADR by Customer Type
df_clean.groupby('customer_type')['adr'].mean().round(2)
```

```python
#4. Estimated Booking Value
df_clean['estimated_booking_value'] = (
    df_clean['adr'] * df_clean['total_nights']
)
df_clean[['adr', 'total_nights', 'estimated_booking_value']].head(10)
print(df_clean['estimated_booking_value'].sum())
```

```python
#5. Total Estimated Booking Value by Hotel
df_clean.groupby('hotel')['estimated_booking_value'].sum().round(2)
```

```python
#6. Monthly Estimated Booking Value
monthly_revenue = df_clean[df_clean['is_canceled'] == 0].groupby(
    df_clean['arrival_date'].dt.month
)['estimated_booking_value'].sum()

monthly_revenue.round(2)
```

```python
#7. Revenue by Market Segment
df_clean[df_clean['is_canceled'] == 0].groupby(
    'market_segment'
)['estimated_booking_value'].sum().sort_values(ascending=False).round(2)
```
